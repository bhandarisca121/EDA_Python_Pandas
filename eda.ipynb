```python
# Rank all the product lines based on their total revenue 
product_line_revenue =df_updated.groupby(['Product line'])['Total Revenue'].sum()
product_line_rank=product_line_revenue.rank(ascending=False, method='dense').sort_values()

print(product_line_rank)
```

```python
# Which moonth generated the highest slaes?
df_updated['Date'] = pd.to_datetime(df_updated['Date'])
df_updated.groupby(df_updated['Date'].dt.month)['Total Revenue'].sum().sort_values(ascending=False)
```

```python
df_updated[df_updated['Total Revenue']>100].groupby('City')['Total Revenue'].sum().sort_values(ascending=False)
```

```python
# Most transaction by week of the day
df_updated['Date'] = pd.to_datetime(df_updated['Date'])
df_updated.groupby(df_updated['Date'].dt.day_name())['Quantity'].sum().sort_values(ascending=False)
```

```python
# Eeekend vs Weekdays sales
df_updated['Day_Type']=df_updated['Date'].dt.day_name().apply(lambda x: 'Weekend' if x in ['Saturday','Sunday', 'Friday'] else 'Weekday')

df_updated.groupby('Day_Type')['Total Revenue'].sum().sort_values(ascending=False)
```

```python
# Which month has the highest sales

df_updated['Month']=df_updated['Date'].dt.month_name()
df_updated.groupby('Month')['Total Revenue'].sum().sort_values(ascending=False).head(5)
```

```python
df_updated['Year']=df_updated['Date'].dt.year
df_updated.groupby('Year')['Total Revenue'].sum().sort_values(ascending=False).head(5)
```
